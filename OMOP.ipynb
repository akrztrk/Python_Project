```python
import requests
import json

# Define the URL and headers
url = 'http://3.216.111.119:8001/search'
headers = {
    'Content-Type': 'application/json'
}

# Define the payload
payload = {
    "standard_concept": "Non-Standard",
    "topk": 5,
    "vocabulary_id": "ICD10CM",
    "validity": True,
    #"source": "JSL",
    "chunk": "HTLV-1"
}

# Make the GET request
response = requests.get(url, headers=headers, data=json.dumps(payload))

# Check if the request was successful
if response.status_code == 200:
    results = response.json()

    # Initialize a dictionary to store the aggregated results
    aggregated_results = {}

    # Aggregate the results
    for result in results:
        for key, value in result.items():
            if key in aggregated_results:
                aggregated_results[key].append(value)
            else:
                aggregated_results[key] = [value]

    # Print the aggregated results
    for key, value in aggregated_results.items():
        print(f"{key}: {value}")
else:
    print(f"Request failed with status code: {response.status_code}")
    print(response.text)  # Print the response content if not successful
```
